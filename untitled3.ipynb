```python
import pandas as pd
from datetime import datetime

# Load and Display the Dataset
df = pd.read_csv('pandas_functions_data.csv')
print(df.head())

# Get Basic Information
print(df.info())
print(df.describe())

# Select Specific Columns
print(df[['Name', 'Salary']])

# Filter Employees by Age
print(df[df['Age'] > 30])

# Sort Employees by Salary (Highest to Lowest)
print(df.sort_values(by='Salary', ascending=False))

# Find Missing Values
print(df.isnull().sum())

# Group Employees by Department
print(df.groupby('Department').size())
print(df.groupby('Department')['Salary'].mean())

# Add a New Column: Years of Experience
df['Years_of_Experience'] = datetime.now().year - pd.to_datetime(df['Joining_Date']).dt.year
print(df.head())

# Rename and Modify Columns
df.rename(columns={'Salary': 'Annual_Salary'}, inplace=True)
df['Department'] = df['Department'].str.upper()
print(df.head())

# Save the Updated Data to a New CSV File
df.to_csv('updated_pandas_functions_data.csv', index=False)
```

```python
import pandas as pd
from datetime import datetime

# Load and Display the Dataset
df = pd.read_csv('eda_employee_data.csv')
df.head()
```

```python
print(df)
```

```python
# Find Basic Information
print(df.info())
print(f"Rows, Columns: {df.shape}")

# Identify Missing Values
print(df.isnull().sum())

# Handle Missing Values
df['Name'].fillna('Unknown', inplace=True)
df['Salary'].fillna(df['Salary'].mean(), inplace=True)

# Identify and Remove Duplicate Records
print(f"Duplicate Rows: {df.duplicated().sum()}")
df.drop_duplicates(inplace=True)

# Correct Incorrect Values
df.loc[df['Age'] < 0, 'Age'] = df['Age'][df['Age'] > 0].mean()

# View Summary Statistics
print(df.describe())

# Group Employees by Department
print(df.groupby('Department').size())
print(df.groupby('Department')['Salary'].mean())

# Convert Data Types
df['Joining_Date'] = pd.to_datetime(df['Joining_Date'])
df['Age'] = df['Age'].astype(int)

# Save the Cleaned Data to a New CSV File
df.to_csv('cleaned_eda_employee_data.csv', index=False)
```
